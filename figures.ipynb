# Results Visualization and Comparative Analysis

## Objective

This notebook summarizes the final results obtained from:

- Centralized CNN
- FedAvg (IID)
- FedAvg (Non-IID)
- FedProx (Non-IID)

and generates publication-quality figures for the GitHub repository.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Final Experimental Results

```python
methods = [
    'Centralized CNN',
    'FedAvg IID',
    'FedAvg Non-IID',
    'FedProx μ=0.001',
    'FedProx μ=0.01',
    'FedProx μ=0.1'
]

accuracies = [
    98.93,
    99.05,
    71.50,
    82.98,
    85.60,
    73.88
]

results = pd.DataFrame({
    'Method': methods,
    'Accuracy (%)': accuracies
})

results
```

## Figure 1 — Overall Comparison

```python
plt.figure(figsize=(10,5))

plt.bar(methods, accuracies)

plt.ylabel('Accuracy (%)')
plt.title('Comparison of Centralized and Federated Learning Methods')
plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    'final_comparison.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## Figure 2 — Impact of Non-IID Data

```python
labels = [
    'Centralized',
    'FedAvg IID',
    'FedAvg Non-IID'
]

scores = [
    98.93,
    99.05,
    71.50
]

plt.figure(figsize=(8,5))

plt.bar(labels, scores)

plt.ylabel('Accuracy (%)')
plt.title('Impact of Non-IID Data on Federated Learning')

plt.tight_layout()

plt.savefig(
    'noniid_impact.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## Figure 3 — FedProx Recovery Effect

```python
labels = [
    'FedAvg Non-IID',
    'FedProx μ=0.001',
    'FedProx μ=0.01',
    'FedProx μ=0.1'
]

scores = [
    71.50,
    82.98,
    85.60,
    73.88
]

plt.figure(figsize=(8,5))

plt.bar(labels, scores)

plt.ylabel('Accuracy (%)')
plt.title('FedProx Performance under Non-IID Conditions')

plt.tight_layout()

plt.savefig(
    'fedprox_improvement.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## Figure 4 — Results Table

```python
fig, ax = plt.subplots(figsize=(8,3))

ax.axis('off')

table = ax.table(
    cellText=results.values,
    colLabels=results.columns,
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1,1.5)

plt.savefig(
    'results_table.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## Key Findings

1. FedAvg under IID conditions achieved performance comparable to centralized learning.

2. Extreme Non-IID client distributions caused a substantial performance drop:
   - 99.05% → 71.50%

3. FedProx mitigated client drift and improved performance under Non-IID settings.

4. The best FedProx configuration was μ = 0.01:
   - 85.60% accuracy

5. FedProx recovered:
   - +14.10 percentage points compared to FedAvg Non-IID.
